## Fundamentação e justificativa dos modelos de previsão

Para escolher o melhor modelo de previsão a fim de analisar se as moléculas são ou não prontamente biodegradáveis, é necessário que haja a comparação entre modelos, com os hiperparâmetros otimizados, medindo seu desempenho ao final. Foram escolhidos cinco modelos para realizar a categorização binária. Nesta seção explico em qual hipótese se fundamenta cada modelo para entender seu funcionamento. Todas as informações foram tiradas da página de aprendizado supervisionado do *scikit* [3]

#### *K*-Vizinhos Próximos (*K-Nearest Neighbors* -- KNN)
O KNN é um algoritmo baseado na proximidade entre as amostras. Para classificar uma nova molécula, o modelo identifica as \(k\) amostras mais próximas no conjunto de treinamento, considerando uma medida de distância entre seus atributos. A classe prevista é determinada pela votação entre os vizinhos selecionados, podendo atribuir pesos diferentes de acordo com a distância. O algoritmo não constrói um modelo explícito durante o treinamento, pois utiliza diretamente os dados armazenados para realizar as previsões.

#### Árvore de Decisão
A Árvore de Decisão realiza classificações por meio de uma estrutura hierárquica de decisões. O algoritmo divide os dados com base nos atributos que melhor separam as classes, utilizando critérios como a entropia de Shannon. Cada nó representa uma condição sobre determinado atributo, enquanto as folhas representam as classes previstas. Sua principal vantagem é a facilidade de interpretação, pois permite visualizar as condições utilizadas para classificar uma molécula.

#### Floresta Aleatória
A Floresta Aleatória é um algoritmo de aprendizado supervisionado baseado na combinação de múltiplas árvores de decisão. Cada árvore é treinada utilizando diferentes amostras dos dados e subconjuntos aleatórios de atributos. Para realizar a classificação, o modelo combina as previsões das árvores por meio de votação majoritária, selecionando a classe mais frequentemente prevista. Essa abordagem reduz a variância e ajuda a evitar o sobreajuste (overfitting), proporcionando maior capacidade de generalização.

#### Regressão Logística
A Regressão Logística é um modelo de classificação que estima a probabilidade de uma amostra pertencer a determinada classe. Para isso, utiliza uma combinação linear dos atributos de entrada, transformada pela função logística, que produz valores entre 0 e 1. A classificação é realizada a partir de um limiar de decisão aplicado à probabilidade estimada. Apesar do nome, é utilizada para problemas de classificação, sendo especialmente adequada para classificações binárias.

#### Naive Bayes Gaussiano
O Naive Bayes Gaussiano é um algoritmo probabilístico fundamentado no teorema de Bayes. Sua principal hipótese é que os atributos são condicionalmente independentes entre si quando considerada uma determinada classe. Além disso, assume que os valores de cada atributo seguem uma distribuição normal (Gaussiana) dentro de cada classe. A partir dessas distribuições, o modelo calcula a probabilidade de uma molécula pertencer a cada categoria e seleciona aquela com maior probabilidade estimada. Vale ressaltar que mesmo que os atributos não sigam a Gaussiana, este modelo ainda consegue performar bem, sendo válido para a análise.

## Pipeline de modelagem

Antes de realizar a otimização de hiperparâmetros, vale analisar como cada modelo performará com os hiperparâmetros padrão, criado com pipeline, e será feita então uma validação cruzada desses modelos iniciais, antes do optuna. ### Funções

Abaixo é rodado o pipeline de de `KNeighborsClassifier` com o número de vizinhos *k* = 3, valor padrão desse modelo. Para criar os modelos com pipeline foi utilizada a função `Pipeline`, utilizando técnica de redução de dimensionalidade de `PCA` (Análise das Componentes Principais, do inglês) [4] com número de componentes 80% pois reduz a dimensionalidade ainda preservando grande parte da originalidade dos dados. Para melhor desempenho do modelo os dados serão normalizados com o normalizador padrão `StandartScaler`. Este normalizador também será aplicado a todos os modelos por conta da diferença de escalas e amplitudes presente nas definições de diversas características das moléculas.

```python
NUM_VIZINHOS = 3
N_COMPONENTES = 0.8

modelo_knn_pipeline = Pipeline([("scaler", StandardScaler()),
          ("reducer", PCA(n_components=N_COMPONENTES, svd_solver="full")),
          ("clf", KNeighborsClassifier(n_neighbors=NUM_VIZINHOS))])
          
modelo_knn_pipeline.fit(x_treino, y_treino);
```

Abaixo é rodado o pipeline de `LogisticRegression`.

```python
modelo_logreg_pipeline = Pipeline([("scaler", StandardScaler()),
          ("reducer", PCA(n_components= N_COMPONENTES)),
          ("clf", LogisticRegression(max_iter=1000, random_state=SEMENTE_ALEATORIA))])

modelo_logreg_pipeline.fit(x_treino, y_treino);
```

Abaixo é rodado o pipeline de `DecisionTreeClassifier`. 

```python
modelo_arvore_pipeline = Pipeline([
        ("reducer", PCA(n_components= N_COMPONENTES)),
        ("clf", DecisionTreeClassifier(random_state=SEMENTE_ALEATORIA)),
    ])

modelo_arvore_pipeline.fit(x_treino, y_treino)
```

Abaixo é rodado o pipeline de `RandomForestClassifier`. O hiperparâmetro de *bootstrap* (amostragem com reposição) está como verdadeiro pois este garante diminuição na variância do modelo, sendo avaliado como importante ao utilizar este modelo.

```python
modelo_floresta_pipeline = Pipeline([
        ("reducer", PCA(n_components= N_COMPONENTES)),
        ("clf", RandomForestClassifier(bootstrap=True, random_state=SEMENTE_ALEATORIA)),
    ])

modelo_floresta_pipeline.fit(x_treino, y_treino);
```

Abaixo é rodado o pipeline de `BernoulliNB`.

```python
modelo_gaussianNB_pipeline = Pipeline([
        ("reducer", PCA(n_components= N_COMPONENTES)),
        ("clf", GaussianNB()),
    ])

modelo_gaussianNB_pipeline.fit(x_treino, y_treino);
```

Dicionário com a associação entre os nomes dos modelos e os modelos criados, para facilitar a computação dos dados na validação cruzada.

```python
CONSTRUTORES = {
    "Baseline": modelo_baseline_pipeline,
    "KNN": modelo_knn_pipeline,
    "Regressão logística": modelo_logreg_pipeline,
    "Árvore de decisão": modelo_arvore_pipeline,
    "Floresta aleatória": modelo_floresta_pipeline,
    "Naive Bayes": modelo_gaussianNB_pipeline,
}
```

### Validação cruzada inicial

A validação cruzada é utilizada para diminuir o efeito do acaso. Nesse caso, utilizaremos então a função `cross_val_score` para realizar a validação cruzada e `make_scorer` para criar uma função com a métrica utilizada de `average_precision_score`. Para realizar essa validação, o parâmetro cv (estratégia de partição dos dados) está recebendo o `StratifiedGroupKFold` que leva em conta a proporção do target e os grupos similares para evitar vazamento de dados.

```python
grupos_treino = grupos.iloc[indices_treino]

ap = make_scorer(average_precision_score, response_method="predict_proba", pos_label="RB")

cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEMENTE_ALEATORIA)

resultados = {}
for nome, modelo in CONSTRUTORES.items():
    s = cross_val_score(modelo, x_treino, y_treino, groups=grupos_treino, cv=cv, scoring=ap, n_jobs=-1)
    resultados[nome] = {"AP (média)": s.mean(), "Desvio": s.std()}

df_resultados_sem_otimizacao = pd.DataFrame(resultados).T.sort_values("AP (média)", ascending=False)

df_resultados_sem_otimizacao
```

A Regressão Logística foi que melhor performou na métrica utilizada, com um bom razoável, com um desvio padrão mediano. Com essa análise dos pipelines de cada modelo com os hiperparâmetros padrão de cada, o modelo a ser escolhido seria a Regressão Logística, que conseguiu superar, muito bem inclusive, a Floresta Aleatória. Interessante notar que o modelo de *k*-vizinhos próximos performou melhor também que a Floresta Aleatória, que provavelmente indica que o modelo está muito mal otimizado. ### Funções do optuna

As funções abaixo foram retiradas e alteradas do material do Professor Daniel R. Cassar [6].

A função `sugere_pca` abaixo serve para fazer a otimização do PCA em todos os modelos ao ajustar o pipeline no optuna.

```python
def sugere_pca(trial):
    if trial.suggest_categorical("usar_pca", [True, False]):
        return PCA(n_components=trial.suggest_float("variancia_pca", 0.70, 0.99),
                   svd_solver="full")
    return "passthrough"
```

As funções `cria_instancia_` abaixo criam uma instância em optuna com a variação dos hiperparâmetros discutidos anteriormente e retornam o pipeline do modelo para ser utilizado adiante.

```python
# INSTÂNCIA KNN

def cria_instancia_KNN(trial):
    """Cria uma instância do modelo.
    Args:
      trial: objeto tipo Trial do optuna.
    Returns:
      Uma instância do modelo desejado.
    """

    parametros = {
        # espaço de busca
        "n_neighbors": trial.suggest_int("num_vizinhos", 1, 800, log=True),
        "weights": trial.suggest_categorical("peso", ["uniform", "distance"]),
        "metric": trial.suggest_categorical("métrica", ["euclidean", "manhattan", "minkowski"]),
        # argumentos fixos
        "n_jobs": -1,   
    }
        
    modelo = KNeighborsClassifier(**parametros)

    return Pipeline([("scaler", StandardScaler()), ("pca", sugere_pca(trial)), ("classifier", modelo)])

# INSTÂNCIA REGRESSÃO LOGÍSTICA
    
def cria_instancia_LogReg(trial):
    """Cria uma instância do modelo.
    Args:
      trial: objeto tipo Trial do optuna.
    Returns:
      Uma instância do modelo desejado.
    """

    parametros = {
        # espaço de busca
        "C": trial.suggest_float("C", 0.001, 1000, log=True),
        "penalty": trial.suggest_categorical("penalty", ["l1", "l2"]),
        "solver": trial.suggest_categorical("solver", ["liblinear", "saga"]),
        "class_weight": trial.suggest_categorical("class_weight", [None, "balanced"]),
        "max_iter": 5000,
        # argumentos fixos
        "n_jobs": -1,
        "random_state": SEMENTE_ALEATORIA,
    }

    modelo = LogisticRegression(**parametros)

    return Pipeline([("scaler", StandardScaler()), ("pca", sugere_pca(trial)), ("classifier", modelo)])

# INSTÂNCIA ÁRVORE ALEATÓRIA

def cria_instancia_arvore(trial):
    """Cria uma instância do modelo.
    Args:
      trial: objeto tipo Trial do optuna.
    Returns:
      Uma instância do modelo desejado.
    """


    parametros = {
        # espaço de busca
        "criterion": trial.suggest_categorical("critério", ["gini", "entropy", "log_loss"]),
        "splitter": trial.suggest_categorical("estrategia_divisao", ["best", "random"]),
        "max_depth": trial.suggest_int("profundidade_max", 2, 300),
        "min_samples_split": trial.suggest_int("min_exemplos_split", 2, 30),
        "min_samples_leaf": trial.suggest_int("min_exemplos_folha", 1, 20),
        "max_features": trial.suggest_categorical("max_atributos", [None, "sqrt", "log2"]),
        "max_leaf_nodes": trial.suggest_int("max_nos_folha", 10, 800, log=True),
        "min_impurity_decrease": trial.suggest_float("reducao_impureza_min", 0.0, 0.1),
        # não recebe n_jobs ou random_state

    }

    modelo = DecisionTreeClassifier(**parametros)

    return Pipeline([("scaler", StandardScaler()), ("pca", sugere_pca(trial)), ("classifier", modelo)])

# INSTÂNCIA FLORESTA ALEATÓRIA

def cria_instancia_floresta(trial):
    """Cria uma instância do modelo.
    Args:
      trial: objeto tipo Trial do optuna.
    Returns:
      Uma instância do modelo desejado.
    """

    parametros = {
        # espaço de busca
        "n_estimators": trial.suggest_int("num_arvores", 100, 2000, log=True),
        "criterion": trial.suggest_categorical("critério", ["gini", "entropy", "log_loss"]),
        "max_depth": trial.suggest_int("profundidade_max", 5, 800),
        "min_samples_split": trial.suggest_int("min_exemplos_split", 2, 30),
        "min_samples_leaf": trial.suggest_int("min_exemplos_folha", 1, 20),
        "max_features": trial.suggest_float("num_max_atributos", 0.1, 1.0),
        "class_weight": trial.suggest_categorical("peso_classes", [None, "balanced"]),
        "max_samples": trial.suggest_float("fracao_amostras", 0.3, 1.0),
        # argumentos fixos
        "bootstrap": True,
        "n_jobs": -1,
        "random_state": SEMENTE_ALEATORIA,
    }

    modelo = RandomForestClassifier(**parametros)

    return Pipeline([("scaler", StandardScaler()), ("pca", sugere_pca(trial)), ("classifier", modelo)])

# INSTÂNCIA GAUSSNB

def cria_instancia_gaussNB(trial):
    """Cria uma instância do modelo.
    Args:
      trial: objeto tipo Trial do optuna.
    Returns:
      Uma instância do modelo desejado.
    """

    parametros = {
        # espaço de busca
        "var_smoothing": trial.suggest_float("suavizacao_variancia", 1e-12, 1e-6, log=True),
        # não recebe n_jobs ou random_state
    }

    modelo = GaussianNB(**parametros)

    return Pipeline([("scaler", StandardScaler()), ("pca", sugere_pca(trial)), ("classifier", modelo)])
```

Dicionário para não reescrever a função objetivo repetidamente.

```python
INSTANCIAS = {
    "KNN": cria_instancia_KNN,
    "Regressão logística": cria_instancia_LogReg,
    "Árvore de decisão": cria_instancia_arvore,
    "Floresta aleatória": cria_instancia_floresta,
    "Naive Bayes Gaussiano": cria_instancia_gaussNB,
}
```

A `funcao_objetivo` tem como função computar a métrica de desempenho. Neste caso, a métrica de interesse é o *average precision* obtido por validação cruzada. 

```python
def funcao_objetivo(trial, X, y, instancia, grupos):
    """Função objetivo do optuna

    Referencia:
      https://medium.com/@walter_sperat/ using-optuna-with-sklearn-the-right-way-part-1-6b4ad0ab2451
    """

    modelo = INSTANCIAS[instancia](trial)
    
    cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEMENTE_ALEATORIA)

    metricas = cross_val_score(
        modelo,
        X,
        y,
        scoring="average_precision",
        cv=cv,
        groups=grupos
    )

    return metricas.mean()
```

Dicionário relacionando o modelo e o objeto de estudo do optuna para realizar a validação cruzada.

Criação de um dicionário que relaciona o nome do modelo aos melhores hiperparâmetros de acordo com a otimização realizada pelo optuna. A função `FixedTrial` do optuna fixa um trial com hiperparâmetros específicos, no caso os mesmos visualizados ao rodar cada modelo na seção acima. Por último, é adicionado o modelo baseline para fins comparativos.

```python
CONSTRUTORES = {}

for nome, parametros in ESTUDOS_OPTUNA.items():
    trial = FixedTrial(parametros.best_trial.params)
    CONSTRUTORES[nome] = INSTANCIAS[nome](trial)

CONSTRUTORES["Baseline"] = modelo_baseline_pipeline 
```

A célula abaixo roda a validação cruzada, retomando a separação de grupos inicial, definindo a métrica de AP e definindo a forma de estratificação dos dados com `StratifiedGroupKFold`. Os dados são mostrados em um DataFrame na ordem decrescente.

```python
grupos_treino = grupos.iloc[indices_treino]

ap = make_scorer(average_precision_score, response_method="predict_proba", pos_label="RB")

resultados = {}

for nome, modelo in CONSTRUTORES.items():
    s = cross_val_score(modelo, x_treino, y_treino, groups=grupos_treino, cv=cv, scoring=ap, n_jobs=-1)

    resultados[nome] = {"AP (média)": s.mean(), "Desvio": s.std()}

df_resultados_com_otimizacao = pd.DataFrame(resultados).T.sort_values("AP (média)", ascending=False)
```

Visualizando a diferença entre os modelos antes e após a otimização.

```python
print("Resultados sem otimização de hiperparâmetros:\n", df_resultados_sem_otimizacao, "\n\n")
print("Resultados com otimização de hiperparâmetros:\n", df_resultados_com_otimizacao)
```

A célula abaixo mostra o gráfico boxplot da distribuição de médias AP construído com a IA Copilot.

```python
grupos_treino = grupos.iloc[indices_treino]

prauc = make_scorer(average_precision_score, response_method="predict_proba", pos_label="RB")

cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEMENTE_ALEATORIA)

resultados_cv = {}

for nome, modelo in CONSTRUTORES.items():
    scores = cross_val_score(modelo, x_treino, y_treino, groups=grupos_treino, cv=cv, scoring=prauc, n_jobs=-1)

    resultados_cv[nome] = scores

resultados = pd.DataFrame({
    nome: {"AP (média)": scores.mean(), "Desvio": scores.std()}
    for nome, scores in resultados_cv.items()
}).T.sort_values("AP (média)", ascending=False)

plt.figure(figsize=(12, 6))

plt.boxplot(
    resultados_cv.values(),
    tick_labels=resultados_cv.keys(),
    showmeans=True
)

plt.title("Distribuição do AP na Validação Cruzada")
plt.ylabel("AP")
plt.xlabel("Algoritmo")
plt.xticks(rotation=30)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

A matriz de confusão mostra, em uma classificação binária, os verdadeiros positivos (VP), verdadeiros negativos (VN), falsos positivos (FP) e falsos negativos (FN), permitindo visualizar de forma ampla a performance dos modelos de previsão. A célula abaixo treina os modelos com os hiperparâmetros otimizados do dicionário CONSTRUTORES e retoma cada modelo treinado em outro dicionário.

```python
MODELOS_TREINADOS = {}

for nome, modelo in CONSTRUTORES.items():
    modelo.fit(x_treino, y_treino)

    MODELOS_TREINADOS[nome] = modelo
```

As matrizes de confusão são criadas na célula abaixo com uso do seaborn.

```python
for nome, modelo in MODELOS_TREINADOS.items():

    y_previsao = modelo.predict(x_teste)

    matriz_conf = confusion_matrix(y_teste, y_previsao, labels=ordem_labels)
    df = pd.DataFrame(matriz_conf, ordem_labels, ordem_labels)

    sns.heatmap(df, annot=True, annot_kws={"size": 16})

    plt.title(f"Matriz de confusão do modelo {nome}")
    plt.xlabel("Classe prevista")
    plt.ylabel("Classe verdadeira")
    plt.show()

    print(f"AP do modelo {nome}: {average_precision_score(y_teste, modelo.predict_proba(x_teste)[:, 1], pos_label='RB'):.4f}")
```

## Curvas de Precisão-recall

As curvas de precisão-recall permitem observar como a precisão e a revocação variam de acordo com os diferentes limiares de classificação, possibilitando compreender melhor a capacidade de cada algoritmo em identificar moléculas prontamente biodegradáveis (RB). A utilização dessas curvas é particularmente relevante para o presente estudo devido ao desbalanceamento entre as classes, uma vez que a classe RB representa a classe minoritária e constitui o principal interesse da previsão.

A precisão (Precision) corresponde à proporção de moléculas classificadas como RB que realmente pertencem a essa classe, enquanto a revocação (Recall) representa a proporção de moléculas RB que foram corretamente identificadas pelo modelo. O gráfico abaixo foi feito pela IA ChatGPT.

```python
plt.figure(figsize=(10, 7))

for nome, modelo in MODELOS_TREINADOS.items():

    # Obtém a probabilidade associada à classe RB
    indice_rb = list(modelo.classes_).index("RB")
    probabilidades_rb = modelo.predict_proba(x_teste)[:, indice_rb]

    # Calcula precisão, recall e Average Precision
    precisao, recall, _ = precision_recall_curve(
        y_teste,
        probabilidades_rb,
        pos_label="RB"
    )

    ap = average_precision_score(
        y_teste,
        probabilidades_rb,
        pos_label="RB"
    )

    plt.plot(
        recall,
        precisao,
        label=f"{nome} (AP = {ap:.3f})"
    )

# Linha de referência: proporção de moléculas RB no conjunto de teste
proporcao_rb = (y_teste == "RB").mean()

plt.axhline(
    y=proporcao_rb,
    linestyle="--",
    color="gray",
    label=f"Baseline aleatório (AP = {proporcao_rb:.3f})"
)

plt.title("Curvas de Precisão-Recall dos Modelos")
plt.xlabel("Recall")
plt.ylabel("Precisão")
plt.xlim(0, 1)
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend(loc="best")
plt.tight_layout()
plt.show()
```

As curvas Precision–Recall foram utilizadas para avaliar a capacidade dos modelos de identificar moléculas prontamente biodegradáveis (RB), considerando o equilíbrio entre precision (precisão) e recall (revocação). Essa análise foi escolhida em detrimento do desbalanceamento entre as classes RB e NRB, com maior quantidade de amostras não prontamente biodegradáveis. Nesse contexto, a curva Precision–Recall permite avaliar o desempenho dos classificadores de maneira mais direcionada à identificação da classe minoritária, que representa o principal interesse do estudo.

A precision representa a proporção de moléculas previstas como RB que realmente pertencem a essa classe. Portanto, valores elevados indicam que o modelo apresenta menor quantidade de falsos positivos. Já o recall corresponde à proporção de moléculas efetivamente RB que foram corretamente identificadas pelo classificador. Assim, valores elevados indicam maior capacidade de reconhecer as moléculas prontamente biodegradáveis, reduzindo a ocorrência de falsos negativos. Existe, entretanto, uma relação de compromisso entre essas duas métricas: ao aumentar a sensibilidade do modelo para identificar mais moléculas RB, pode ocorrer também um aumento na quantidade de moléculas NRB classificadas incorretamente como RB, ou seja, falsos positivos.

Para sintetizar o desempenho das curvas, foi utilizada a métrica Average Precision (AP), que resume a relação entre precisão e revocação ao longo dos diferentes limiares de classificação. Os resultados obtidos no conjunto de teste demonstraram desempenho superior dos modelos de Regressão Logística, Floresta Aleatória e KNN, que apresentaram AP de 0,9198, 0,9152 e 0,8992, respectivamente. Esses resultados indicam que os três modelos possuem boa capacidade de distinguir moléculas prontamente biodegradáveis das não prontamente biodegradáveis, mantendo um equilíbrio favorável entre precisão e revocação. A Regressão Logística apresentou o maior AP no conjunto de teste, embora a diferença em relação à Floresta Aleatória tenha sido pequena, indicando desempenhos bastante próximos.

Por outro lado, o Naive Bayes Gaussiano e a Árvore de Decisão apresentaram AP inferiores, de 0,7997 e 0,7880, respectivamente. Embora esses resultados sejam consideravelmente superiores ao baseline, indicam menor capacidade de ordenação das amostras de acordo com sua probabilidade de pertencer à classe RB quando comparados aos três modelos de melhor desempenho. O baseline, construído com o classificador que sempre prevê a classe majoritária, apresentou AP de 0,3396. Esse valor é próximo à proporção de amostras RB no conjunto de teste, demonstrando que os modelos avaliados superaram substancialmente uma estratégia de classificação que não identifica efetivamente as características associadas à biodegradabilidade.

É importante destacar que o melhor desempenho no conjunto de teste não deve ser considerado isoladamente. Na validação cruzada, a Floresta Aleatória apresentou o maior AP médio (0,8947), seguida pela Regressão Logística (0,8717) e pelo KNN (0,8622). Embora a Regressão Logística tenha alcançado o maior resultado no teste, a Floresta Aleatória apresentou desempenho médio superior durante a validação cruzada. Essa diferença evidencia a importância de considerar tanto o desempenho no conjunto de teste quanto a capacidade de generalização observada durante a validação cruzada na escolha do modelo final.

De maneira geral, as curvas Precision–Recall e os respectivos valores de AP demonstram que a Regressão Logística, a Floresta Aleatória e o KNN foram os modelos mais adequados entre os avaliados para a identificação de moléculas prontamente biodegradáveis. A escolha entre eles deve considerar também o comportamento das métricas de precisão e revocação nos diferentes limiares de classificação. Em uma aplicação de triagem de moléculas, por exemplo, um modelo com maior recall pode ser interessante quando o objetivo é identificar a maior quantidade possível de moléculas potencialmente biodegradáveis. Em contrapartida, uma maior precision é desejável quando se busca reduzir a classificação incorreta de moléculas não prontamente biodegradáveis como biodegradáveis. Portanto, a definição do limiar de decisão deve considerar o objetivo prático da aplicação, e não somente o valor global de AP.

## Conclusões e principais aprendizados

O presente trabalho teve como objetivo desenvolver e comparar modelos de aprendizado de máquina capazes de classificar moléculas de acordo com sua biodegradabilidade, identificando aquelas prontamente biodegradáveis (RB) e aquelas não prontamente biodegradáveis (NRB). Para isso, foram utilizados cinco algoritmos de classificação: K-vizinhos próximos (KNN), Regressão Logística, Árvore de Decisão, Floresta Aleatória e Naive Bayes Gaussiano. A avaliação foi realizada com base no conjunto de dados QSAR Biodegradation, composto por 1.055 moléculas descritas por 41 atributos numéricos, utilizando o Average Precision (AP) como principal métrica de desempenho.

Um dos principais aprendizados foi a importância de considerar o desbalanceamento das classes durante a escolha das métricas de avaliação. Como a classe NRB representa a maior parte das observações, a utilização isolada da acurácia poderia resultar em uma interpretação excessivamente otimista do desempenho dos classificadores. Nesse contexto, o Average Precision mostrou-se mais adequado ao objetivo do estudo, pois permite avaliar a capacidade dos modelos de identificar a classe minoritária RB, que calhou de ser a classe mais importante na análise, em diferentes limiares de classificação. A comparação com o baseline também demonstrou a importância dessa abordagem: enquanto o classificador que sempre prevê a classe majoritária apresentou AP médio de 0,3372 na validação cruzada, todos os modelos otimizados alcançaram resultados significativamente superiores.

A otimização dos hiperparâmetros por meio do Optuna foi outro fator fundamental para o desempenho dos modelos. A comparação entre os resultados anteriores e posteriores à otimização mostrou melhorias expressivas para todos os algoritmos. Destaca-se a Floresta Aleatória, cujo AP médio na validação cruzada aumentou de 0,6977 para 0,8947. A Árvore de Decisão também apresentou uma evolução significativa, passando de 0,4626 para 0,7612. Esses resultados evidenciam a importância de otimizar os hiperparâmetros de modelos de previsão antes de realizar a escolha de qual modelo utilizar para problemas específicos.

Após a otimização, a Floresta Aleatória apresentou o melhor desempenho médio na validação cruzada, com AP de 0,8947 e desvio padrão de 0,0198, indicando resultados elevados e relativamente consistentes entre as partições avaliadas. A Regressão Logística e o KNN também apresentaram desempenhos competitivos, com AP médios de 0,8717 e 0,8622, respectivamente. No conjunto de teste, entretanto, a Regressão Logística alcançou o maior AP, de 0,9190, seguida pela Floresta Aleatória, com 0,9152, e pelo KNN, com 0,8992. A proximidade entre os resultados desses três modelos demonstra que diferentes abordagens de classificação foram capazes de identificar padrões relevantes nos descritores moleculares. A diferença entre o melhor resultado na validação cruzada e no teste também reforça a importância de avaliar os modelos em diferentes partições, evitando que a escolha final seja determinada exclusivamente por um único conjunto de dados.

Em contrapartida, o Naive Bayes Gaussiano e a Árvore de Decisão apresentaram resultados inferiores aos três modelos de maior desempenho, com AP de 0,7997 e 0,8536 no conjunto de teste, respectivamente. Ainda assim, ambos superaram consideravelmente o baseline. Esses resultados demonstram que diferentes algoritmos podem apresentar capacidades distintas para representar as relações entre os atributos moleculares e a biodegradabilidade. Em particular, o desempenho superior da Floresta Aleatória em relação à Árvore de Decisão sugere que a combinação de múltiplas árvores pode representar melhor os padrões presentes nesse conjunto de dados do que a utilização de uma única estrutura de decisão.

A análise das curvas Precision–Recall e das matrizes de confusão também permitiu compreender que a escolha de um modelo não deve considerar apenas sua capacidade geral de discriminação. É necessário avaliar o equilíbrio entre precisão e revocação e, consequentemente, os tipos de erros cometidos. No contexto deste trabalho, os falsos positivos são particularmente relevantes, pois representam moléculas não prontamente biodegradáveis classificadas incorretamente como biodegradáveis. Assim, embora a Regressão Logística tenha apresentado o maior AP no teste, a escolha do modelo para uma aplicação prática também deve considerar o custo associado a cada tipo de erro e o limiar de classificação mais adequado.

Outro aprendizado importante foi compreender que técnicas de pré-processamento e redução de dimensionalidade não devem ser aplicadas indiscriminadamente. A normalização é especialmente relevante para algoritmos sensíveis às escalas dos atributos, como KNN e Regressão Logística. Já a utilização do PCA foi avaliada durante a otimização, permitindo que o próprio processo identificasse quando a redução de dimensionalidade era vantajosa. Nos três modelos de melhor desempenho, as configurações selecionadas pelo Optuna não utilizaram PCA, indicando que, para esse conjunto de dados e esses algoritmos, preservar os descritores originais foi mais adequado do que reduzir sua dimensionalidade.

Por fim, a utilização da validação cruzada estratificada e agrupada foi importante para tornar a avaliação mais confiável, considerando simultaneamente a proporção das classes e a presença de observações com descritores muito semelhantes. Essa estratégia buscou reduzir o risco de vazamento de informações entre os conjuntos de treinamento e validação, tornando a comparação entre os modelos mais criteriosa.

De maneira geral, os resultados indicam que a Regressão Logística e a Floresta Aleatória foram as alternativas mais competitivas para a classificação da biodegradabilidade molecular. A primeira apresentou o maior Average Precision no conjunto de teste, enquanto a segunda obteve o melhor desempenho médio na validação cruzada. Portanto, ambas se destacam como candidatas para a aplicação proposta, sem que a pequena diferença entre seus resultados permita afirmar uma superioridade definitiva de uma sobre a outra. Mais amplamente, o trabalho demonstrou a importância da exploração dos dados, da seleção adequada das métricas, da otimização dos hiperparâmetros e da avaliação por diferentes estratégias para a construção de modelos preditivos. Os classificadores desenvolvidos podem contribuir para a triagem inicial de moléculas com potencial biodegradável, embora suas previsões não substituam a realização de ensaios experimentais para confirmar o comportamento de biodegradação.
